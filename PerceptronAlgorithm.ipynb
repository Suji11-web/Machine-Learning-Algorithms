# Perceptron Algorithm

**Required Libraries**

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')
```

**Random Dataset Generation**

```python
from sklearn import datasets
x,y=datasets.make_blobs(n_samples=200,n_features=2,centers=[[2,2],[10,10]],cluster_std=3,random_state=42)
```

**Plotting the dataset with random decision boundary**

```python
def plot_line(W,w0,x,y):
  min=x[:,0].min()
  max=x[:,0].max()
  w1=W[0]
  w2=W[1]
  slope=-w1/w2
  intercept=w0/w2
  x_values=np.linspace(min,max,100)
  y_values=slope*x_values+intercept
  sns.scatterplot(x=x[:,0], y=x[:,1], hue=y)
  plt.plot(x_values, y_values, 'r')
  plt.show()
```

```python
W=np.array([0.8,-0.8])
w0=0.92
plot_line(W,w0,x,y)
```

**Algorithm**

```python
def distance_from_line(W,w0,x):
  distance=np.dot(W,x)+w0
  return distance
```

```python
def perceptron(W,w0,x,y,iter):
  for i in range(iter):
    misclassification=0
    plot_line(W,w0,x,y)
    for xi,yi in zip(x,y):
      distance=distance_from_line(W,w0,xi)
      if np.sign(distance)!=yi:
        misclassification+=1
        W=W+xi*yi
        w0+=yi
    if misclassification==0:
      return W,w0
  return W,w0
```

```python
perceptron(W,w0,x,y,10)
```
