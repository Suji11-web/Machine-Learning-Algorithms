# Linear Regression Scratch Code

**Intution**  We use Mean Squared Error as a loss function. And Gradient Descent to optimize the loss function (Finding Minima)

*   Loss function L(W,w0)=(1/n) Σ(y-y^)**2
*   y^=(W**T)*X+w0

*   Initialize random weights
*   Find derivative of the Loss function


*   Updated the weights with the derivative
*   W=W-η((2/n)Σ(y^-y)x)  |  w0=w0-η((2/n)Σ(y^-y)







**Required Libraries**

```python
import numpy as np
import pandas as pd
```

**Predict value**

```python
def predict(X,weights):
  return np.dot(X,weights)
```

**Loss Function**

```python
def loss(X,Y,weights):
  y_hat=predict(X,weights)
  loss=np.mean(Y-y_hat)**2
  return loss
```

**Gradient of the Loss Function**

```python
def gradient(X,Y,weights):
  y_hat=predict(X,weights)
  grad=(2/len(Y))*(np.dot(X.T,(y_hat-Y)))
  return grad
```

**Gradient Descent and Updating Weights**

```python
def gradient_descent(X,Y,learning_rate=0.1,iter=100):
  n,d=X.shape
  error_list=[]

  #Initialization
  W=np.random.rand(d)
  for i in range(iter):
    e=loss(X,Y,W)
    error_list.append(e)

    #Calculating derivative
    dW=gradient(X,Y,W)

    #Updating weights
    W=W-learning_rate*dW
  return W,error_list
```
